```python
from pathlib import Path
import json
import sys

server_dir = Path.cwd()
if not (server_dir / "lib" / "arm.py").is_file():
    server_dir /= "OT2_Server_2026-06-11"
if not (server_dir / "lib" / "arm.py").is_file():
    raise FileNotFoundError("Open Jupyter in the OT-2 protocol folder.")

sys.path.insert(0, str(server_dir / "lib"))
from arm import Arm
from compression_plan import resolve_test_locations

robot = json.loads((server_dir / "robot.json").read_text(encoding="utf-8"))

xArm = Arm(
    coupons=robot["coupons"], rings=robot["rings"],
    discards=robot["discard"], camera_box_open=robot["camera_box_open"], home=True
)
```
